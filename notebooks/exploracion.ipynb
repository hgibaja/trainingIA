# Exploración del dataset

Antes de entrenar, siempre conviene **mirar los datos**: cuántos hay, cómo son y si las clases están equilibradas.

Dataset: [AI-Lab-Makerere/beans](https://huggingface.co/datasets/AI-Lab-Makerere/beans) — fotos de hojas de frijol en 3 clases.

```python
import sys
sys.path.append("../src")  # para poder importar nuestro código

from collections import Counter
import matplotlib.pyplot as plt
from pipeline import cargar_dataset

ds, etiquetas = cargar_dataset()
ds
```

## ¿Cuántas imágenes hay de cada clase?

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
for ax, particion in zip(axes, ["train", "validation", "test"]):
    conteo = Counter(ds[particion]["labels"])
    ax.bar(etiquetas, [conteo[i] for i in range(len(etiquetas))])
    ax.set_title(f"{particion} ({len(ds[particion])} imágenes)")
plt.tight_layout()
```

## Ejemplos de cada clase

```python
fig, axes = plt.subplots(len(etiquetas), 4, figsize=(12, 9))
for clase, nombre in enumerate(etiquetas):
    ejemplos = ds["train"].filter(lambda e: e["labels"] == clase).select(range(4))
    for ax, ejemplo in zip(axes[clase], ejemplos):
        ax.imshow(ejemplo["image"])
        ax.set_title(nombre)
        ax.axis("off")
plt.tight_layout()
```
